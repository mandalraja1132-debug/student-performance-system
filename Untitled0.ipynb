```python
import pandas as pd

# 1. Load departments
departments = pd.read_csv('/content/departments.csv')
print('Departments loaded. Shape:', departments.shape)

# 2. Load students
students = pd.read_csv('/content/students.csv')
print('Students loaded. Shape:', students.shape)

# 3. Load subjects
subjects = pd.read_csv('/content/subjects.csv')
print('Subjects loaded. Shape:', subjects.shape)

# 4. Load marks
marks = pd.read_csv('/content/marks.csv')
print('Marks loaded. Shape:', marks.shape)
```

```python
# 5. Convert marks to numeric and clean duplicates
marks['marks_obtained'] = pd.to_numeric(marks['marks_obtained'], errors='coerce')
marks = marks.drop_duplicates(subset=['student_id', 'subject_id'])

# Fill missing marks with the average mark for that subject
marks['marks_obtained'] = marks.groupby('subject_id')['marks_obtained'].transform(
    lambda series: series.fillna(series.mean())
)

print('Data cleaning and preprocessing completed!')
print('Total records after cleaning:', len(marks))
```

```python
# 6. Merge preprocessed datasets together
merged_marks = marks.merge(students, on='student_id', how='inner')
merged_marks = merged_marks.merge(departments, on='department_id', how='left')
merged_marks = merged_marks.merge(subjects, on='subject_id', how='inner')

print('Datasets merged successfully!')
print('Columns in merged dataset:', merged_marks.columns.tolist())
display(merged_marks.head())
```

```python
# 7. Build Student Report DataFrame
pivot_report = merged_marks.pivot_table(
    index=['student_id', 'roll_no', 'student_name', 'department_name'],
    columns='subject_name',
    values='marks_obtained',
    aggfunc='mean'
).reset_index()

# Calculate totals, percentages, and results
subject_cols = ['Python Programming', 'Database Systems', 'Statistics', 'Data Structures', 'Machine Learning', 'Data Visualization']
pivot_report['total_marks'] = pivot_report[subject_cols].sum(axis=1)
pivot_report['percentage'] = pivot_report[subject_cols].mean(axis=1)

def determine_grade(pct):
    if pct >= 85: return 'A+'
    elif pct >= 75: return 'A'
    elif pct >= 65: return 'B'
    elif pct >= 50: return 'C'
    elif pct >= 40: return 'D'
    else: return 'F'

pivot_report['grade'] = pivot_report['percentage'].apply(determine_grade)
pivot_report['result'] = pivot_report[subject_cols].apply(
    lambda row: 'Pass' if (row >= 40).all() else 'Fail', axis=1
)

# Calculate ranks based on percentage
pivot_report['rank'] = pivot_report['percentage'].rank(ascending=False, method='min').astype(int)

# Sort by rank
pivot_report = pivot_report.sort_values(by='rank')

print('Student Report successfully built!')
display(pivot_report.head())
```

```python
# 8. Subject-wise Analysis
print('=== SUBJECT-WISE ANALYSIS ===')
subject_stats = merged_marks.groupby('subject_name')['marks_obtained'].agg(['mean', 'min', 'max']).reset_index()

# Calculate Pass Rate per Subject (vectorized, avoiding groupby.apply deprecation warning)
merged_marks['passed'] = merged_marks['marks_obtained'] >= 40
pass_rates = merged_marks.groupby('subject_name')['passed'].mean().reset_index(name='pass_rate_pct')
pass_rates['pass_rate_pct'] = pass_rates['pass_rate_pct'] * 100

subject_analysis = subject_stats.merge(pass_rates, on='subject_name')
display(subject_analysis.round(2))

# Department-wise Analysis
print('\n=== DEPARTMENT-WISE ANALYSIS ===')
dept_stats = pivot_report.groupby('department_name')['percentage'].mean().reset_index(name='average_percentage')
display(dept_stats.round(2))
```

```python
# Chart 1: Subject-wise Average Marks
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 4))
sns.barplot(data=subject_analysis, x='subject_name', y='mean', hue='subject_name', palette='muted', legend=False)
plt.title('Subject-wise Average Marks')
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig('/content/average_marks_by_subject.png')
plt.show()
```

```python
# Chart 2: Grade Distribution (Pie Chart)
import matplotlib.pyplot as plt

# Count grades and filter out those with 0 students
grade_counts = pivot_report['grade'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    grade_counts.values,
    labels=grade_counts.index,
    autopct='%1.1f%%',
    startangle=140
)
plt.title('Overall Grade Distribution')
plt.tight_layout()
plt.savefig('/content/grade_distribution.png')
plt.show()
```

```python
# Chart 3: Distribution of Marks Obtained (Histogram)
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 4))
sns.histplot(data=merged_marks, x='marks_obtained', kde=True, bins=10, color='skyblue')
plt.title('Distribution of Marks Obtained')
plt.tight_layout()
plt.savefig('/content/marks_histogram.png')
plt.show()
```

```python
# Chart 4: Student Percentage Distribution by Department (Boxplot)
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.boxplot(data=pivot_report, x='department_name', y='percentage', hue='department_name', palette='Set3', legend=False)
plt.title('Percentage Distribution by Department')
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig('/content/percentage_by_department_boxplot.png')
plt.show()
```

```python
# 9. Show Topper List (Top 10 Students)
print('=== TOP 10 TOPPERS ===')
toppers = pivot_report.head(10)[['rank', 'roll_no', 'student_name', 'percentage', 'grade', 'result']]
display(toppers.to_string(index=False))

# 10. Export Student Report to CSV
output_path = '/content/student_report.csv'
pivot_report.to_csv(output_path, index=False)
print(f'\nStudent report successfully exported to: {output_path}')
```
